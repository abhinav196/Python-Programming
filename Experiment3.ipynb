1.Factorial of a Given Number



```python
n = int(input("Enter a number: "))
fact = 1

for i in range(1, n + 1):
    fact *= i

print("Factorial =", fact)
```

2.Check Armstrong Number

```python
num = int(input("Enter a number: "))
order = len(str(num))
temp = num
sum = 0

while temp > 0:
    digit = temp % 10
    sum += digit ** order
    temp //= 10

if sum == num:
    print("Armstrong Number")
else:
    print("Not an Armstrong Number")
```

3.Fibonacci Series up to Given Term

```python
n = int(input("Enter number of terms: "))
a, b = 0, 1

for i in range(n):
    print(a, end=" ")
    a, b = b, a + b
```

4.Check Prime Number

```python
num = int(input("Enter a number: "))

if num > 1:
    for i in range(2, num):
        if num % i == 0:
            print("Not Prime")
            break
    else:
        print("Prime Number")
else:
    print("Not Prime")
```

5.Check Palindrome Number

```python
num = input("Enter a number: ")

if num == num[::-1]:
    print("Palindrome")
else:
    print("Not Palindrome")
```

6.Sum of Digits

```python
num = int(input("Enter a number: "))
sum = 0

while num > 0:
    sum += num % 10
    num //= 10

print("Sum of digits =", sum)
```

7. Count Numbers Divisible by 5 or 7 (1 to 100)


```python
count = 0

for i in range(1, 101):
    if i % 5 == 0 or i % 7 == 0:
        print(i, end=" ")
        count += 1

print("\nTotal count =", count)
```

9.Print Table of a Given Number

```python
n = int(input("Enter a number: "))

for i in range(1, 11):
    print(n, "*", i, "=", n * i)
```

10.Pattern Program

```python
n = 5

for i in range(n, 0, -1):
  for j in range(1, i + 1):
    print(j, end="")
    for k in range(2 * (n - i)):
      print("*", end="")
      for j in range(i, 0, -1):
        print(j, end="")

    print()
```

11.Sum of Series (1 + 1/2 + 1/3 + ... + 1/n)

```python
n = int(input("Enter value of n: "))
sum = 0

for i in range(1, n + 1):
    sum += 1/i

print("Sum of series =", sum)
```
