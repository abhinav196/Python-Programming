1. Count and display number of capital letters in a string

```python
s = input("Enter a string: ")
count = 0

for ch in s:
    if ch.isupper():
        count += 1

print("Number of capital letters:", count)
```

2. Count total number of vowels in a string

```python
s = input("Enter a string: ")
vowels = "aeiouAEIOU"
count = 0

for ch in s:
    if ch in vowels:
        count += 1

print("Total number of vowels:", count)
```

3. Input a sentence and print words in separate lines

```python
sentence = input("Enter a sentence: ")

words = sentence.split()

for word in words:
    print(word)
```

4. Count number of times substring occurs (Left to Right)

```python
string = input("Enter string: ")
substring = input("Enter substring: ")

count = 0

for i in range(len(string) - len(substring) + 1):
    if string[i:i+len(substring)] == substring:
        count += 1

print("Number of occurrences:", count)
```

5. Count occurrences of each alphabet (case insensitive)

```python
s = input("Enter a string: ")
s = s.upper()

result = {}

for ch in s:
    if ch.isalpha():
        if ch in result:
            result[ch] += 1
        else:
            result[ch] = 1

for key in sorted(result):
    print(result[key], key)
```

6. Count number of unique words using sets

```python
sentence = input("Enter a sentence: ")

words = sentence.split()
unique_words = set(words)

print("Number of unique words:", len(unique_words))
```

7. Create two sets of fruits and perform operations

```python
n = int(input("Enter number of fruits in each set: "))

s1 = set()
s2 = set()

print("Enter fruits for Set 1:")
for i in range(n):
    s1.add(input())

print("Enter fruits for Set 2:")
for i in range(n):
    s2.add(input())

print("Common fruits:", s1 & s2)

print("Fruits only in Set 1:", s1 - s2)

print("Total unique fruits:", len(s1 | s2))
```

8. Apply set operations

```python
S1 = {"Red", "yellow", "orange", "blue"}
S2 = {"violet", "blue", "purple"}

print("Union:", S1 | S2)
print("Intersection:", S1 & S2)
print("Difference (S1 - S2):", S1 - S2)
print("Difference (S2 - S1):", S2 - S1)
print("Symmetric Difference:", S1 ^ S2)
```
