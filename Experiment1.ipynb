5.Length of Hypotenuse using Pythagoras Theorem

```python
import math

a = float(input("Enter side a: "))
b = float(input("Enter side b: "))

c = math.sqrt(a*a + b*b)

print("Length of hypotenuse:", c)
```

7.Area of Triangle (Heron’s Formula)

```python
import math

a = float(input("Enter side 1: "))
b = float(input("Enter side 2: "))
c = float(input("Enter side 3: "))

s = (a + b + c) / 2
area = math.sqrt(s * (s - a) * (s - b) * (s - c))

print("Area of triangle:", area)
```

8.Convert Seconds into Hours, Minutes, Seconds

```python
seconds = int(input("Enter total seconds: "))

hours = seconds // 3600
minutes = (seconds % 3600) // 60
remaining_seconds = seconds % 60

print("Hours:", hours)
print("Minutes:", minutes)
print("Seconds:", remaining_seconds)
```

9.Swap Two Numbers Without Extra Variable

```python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

a, b = b, a

print("After swapping:")
print("a =", a)
print("b =", b)
```

10.Sum of First n Natural Numbers

```python
n = int(input("Enter n: "))

sum_n = n * (n + 1) // 2

print("Sum of first", n, "natural numbers is:", sum_n)
```

11.Truth Table for Bitwise Operators (&, |, ^)

```python
print("A B  A&B  A|B  A^B")
for a in [0, 1]:
    for b in [0, 1]:
        print(a, b, "  ", a & b, "   ", a | b, "   ", a ^ b)
```

12.Left Shift and Right Shift

```python
num = int(input("Enter a number: "))
shift = int(input("Enter number of positions to shift: "))

print("Left Shift:", num << shift)
print("Right Shift:", num >> shift)
```

13.Membership Operator

```python
sequence = (10, 20, 56, 78, 89)

num = int(input("Enter number to check: "))

if num in sequence:
    print("Number is present in sequence.")
else:
    print("Number is NOT present in sequence.")
```
