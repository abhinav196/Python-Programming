1. Function to find Maximum and Minimum (Without built-in functions)

```python
def find_max_min(numbers):
    max_val = numbers[0]
    min_val = numbers[0]

    for num in numbers:
        if num > max_val:
            max_val = num
        if num < min_val:
            min_val = num

    return max_val, min_val


nums = [10, 6, 8, 90, 12, 56]
print(find_max_min(nums))

```

2. Function to return sum of cubes of positive integers smaller than given number

```python
def sum_of_cubes(n):
    total = 0
    for i in range(1, n):
        total += i ** 3
    return total


print(sum_of_cubes(5))  # 1^3 + 2^3 + 3^3 + 4^3
```

3. Print 1 to n using recursion (No loop)

```python
def print_numbers(n):
    if n == 0:
        return
    print_numbers(n - 1)
    print(n)


print_numbers(5)
```

4. Recursive Fibonacci series up to n terms

```python
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)

def print_fibonacci_terms(n):
    for i in range(n):
        print(fibonacci(i), end=" ")


print_fibonacci_terms(7)
```

5. Lambda function to find volume of cone

```python
import math

volume_cone = lambda r, h: (1/3) * math.pi * r**2 * h

print(volume_cone(3, 5))
```

6. Lambda function to return tuple of max and min

```python
numbers = [10, 6, 8, 90, 12, 56]

max_min = lambda lst: (max(lst), min(lst))

print(max_min(numbers))
```

7. Functions explaining arguments
a. Keyword Argument

```python
def student(name, age):
    print("Name:", name)
    print("Age:", age)

student(age=20, name="Rahul")
```

b. Default Argument

```python
def greet(name="Guest"):
    print("Hello", name)

greet()
greet("Ankit")
```

c.Variable-length Argument


```python
def total_sum(*numbers):
    total = 0
    for num in numbers:
        total += num
    print("Sum:", total)

total_sum(10, 20, 30)
```

8. Check whether all dictionary values are same using lambda

```python
data = {'a': 10, 'b': 10, 'c': 10}

check_same = lambda d: len(set(d.values())) == 1

print(check_same(data))
```

9. Create dictionary using two lists

```python
list1 = ['a', 'b', 'c']
list2 = [1, 2, 3]

dictionary = {}

for i in range(len(list1)):
    dictionary[list1[i]] = list2[i]

print(dictionary)
```
